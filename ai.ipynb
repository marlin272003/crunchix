```python
from textblob import TextBlob

text = """
I really love this project.
It is amazing and very useful.
"""

blob = TextBlob(text)

polarity = blob.sentiment.polarity

print("Polarity:", polarity)

if polarity > 0:
    print("Positive")
elif polarity < 0:
    print("Negative")
else:
    print("Neutral")
```
