```python
from textblob import TextBlob
import pandas as pd

def analyze_file_sentiment(file_name):
    results = []

    try:
        # فتح وقراءة الملف
        with open(file_name, 'r', encoding='utf-8') as file:
            lines = file.readlines()

        for line in lines:
            text = line.strip()
            if text:  # التأكد أن السطر ليس فارغاً
                # التحليل باستخدام TextBlob
                analysis = TextBlob(text)
                polarity = analysis.sentiment.polarity

                # تصنيف النتيجة
                if polarity > 0:
                    sentiment = "Positive"
                elif polarity < 0:
                    sentiment = "Negative"
                else:
                    sentiment = "Neutral"

                results.append({"Text": text, "Sentiment": sentiment, "Score": round(polarity, 2)})

        # عرض النتائج بشكل منظم في جدول
        df = pd.DataFrame(results)
        return df

    except FileNotFoundError:
        print(f"خطأ: الملف '{file_name}' غير موجود. تأكدي من رفعه على Colab أولاً.")
        return None

# --- تشغيل الكود ---
# استبدلي 'data.txt' باسم ملفك الحقيقي
filename = 'data.txt'
final_results = analyze_file_sentiment(filename)

if final_results is not None:
    print("تم تحليل الملف بنجاح! إليك النتائج:")
    display(final_results)
```
